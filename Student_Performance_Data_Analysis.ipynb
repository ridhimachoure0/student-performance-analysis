```python
import pandas as pd
```

```python
data = {"student" : ["S1","S2","S3","S4","S5","S6","S7","S8","S9","S10"],"Study_Hours" : [2,4,5,6,3,7,8,1,5,6], "Attendance" : [70,78,85,90,75,92,95,65,82,88],"Score" : [52,63,70,78,58,85,91,45,72,80]}

df = pd.DataFrame(data)

df
```

```python
print (type(df))
print (df.columns)
```

```python
df["Score"].mean()

df["Score"].median()

df["Score"].std()
```

```python
df[["Study_Hours","Score"]].corr()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df["Study_Hours"],df["Score"])

plt.xlabel("Study_Hours")
plt.ylabel("Score")
plt.title("study_Hours vs Score")

plt.show()
```

```python
plt.scatter(df["Attendance"], df["Score"])

plt.xlabel("Attendance (%)")
plt.ylabel("Score")
plt.title("Attendance vs Score")

plt.show()
```

```python
plt.hist(df["Score"], bins=5)

plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Scores")

plt.show()
```
